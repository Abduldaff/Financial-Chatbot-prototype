```python
companies = ['Microsoft','Tesla','Apple']
bcg_df = df[df['Company'].isin(companies)].copy()

df.head()
```

```python
bcg_df = bcg_df[bcg_df['Year'].between(2023,2025)].copy()
bcg_df.sort_values(['Company','Year'])
```

```python
financial_df = bcg_df[
    [
        "Company",
        "Year",
        "Source",
        "Total Revenue in USD",
        "Net Income in USD",
        "Total Assets in USD",
        "Total Liablilities in USD",
        "Cash Flow from operating activities in USD"
    ]].copy()
financial_df.head()
```

```python
financial_df = financial_df.rename(columns={
    "Total Revenue in USD": "Total Revenue",
    "Net Income in USD": "Net Income",
    "Total Assets in USD": "Total Assets",
    "Total Liablilities in USD": "Total Liabilities",
    "Cash Flow from operating activities in USD": "Operating Cash Flow"
})

financial_df.head()
```

```python
print(financial_df.columns.tolist())
```

```python
print(financial_df["Company"].unique())
print(sorted(financial_df["Year"].unique()))
```

```python
financial_df.sort_values(['Company','Year'])
```

```python
financial_df["Revenue Growth (%)"] = (
    financial_df.groupby("Company")["Total Revenue"].pct_change() * 100
)
financial_df[["Company", "Year", "Total Revenue", "Revenue Growth (%)"]].round(2)
```

```python
financial_df["Net Income Growth (%)"] = (
    financial_df.groupby("Company")["Net Income"].pct_change() * 100
)
```

```python
financial_df["Profit Margin (%)"] = (
    financial_df["Net Income"]/ financial_df["Total Revenue"]
) * 100
```

```python
financial_df["Asset Growth (%)"] = (
    financial_df.groupby("Company")["Total Assets"].pct_change() * 100
)
```

```python
financial_df["Liability Growth (%)"] = (
    financial_df.groupby("Company")["Total Liabilities"].pct_change() * 100
)
```

```python
financial_df["Operating Cash Flow Growth (%)"] = (
    financial_df.groupby("Company")["Operating Cash Flow"].pct_change() * 100
)
```

```python
financial_df.round(2)
```

```python
plt.figure(figsize=(10, 6))
for company in financial_df["Company"].unique():
    data = financial_df[
        financial_df["Company"] == company]
    plt.plot(
        data["Year"],
        data["Total Revenue"],
        marker="o",
        label=company
    )
plt.title("Revenue Trend: 2023–2025")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
plt.show()
```

## Financial Analysis Findings
The analysis compares the financial performance of Microsoft, Tesla, and Apple between 2023 and 2024.

### Microsoft
Microsoft recorded strong financial growth during the period. Total revenue increased by 15.67%, while net income increased by 21.80%. Its profit margin improved from 34.15% in 2023 to 35.96% in 2024. Total assets increased by 24.32%, and operating cash flow increased by 35.36%, indicating strong growth in operating cash generation.

### Tesla
Tesla's revenue increased marginally by 0.95% between 2023 and 2024. However, net income decreased by 52.23%, resulting in a reduction in profit margin from 15.47% to 7.32%. Total assets increased by 14.49%, while operating cash flow increased by 12.58%. The results show a significant difference between revenue growth and profitability during the period.

### Apple
Apple's revenue increased by 2.02%, while net income decreased by 3.36%. Profit margin declined from 25.31% in 2023 to 23.97% in 2024. Total assets increased by 3.52%, liabilities increased by 6.06%, and operating cash flow increased by 6.98%.

### Overall Observation
The analysis shows different financial trends across the three companies. Microsoft experienced growth in revenue, net income, profit margin, assets, and operating cash flow. Tesla experienced limited revenue growth alongside a substantial decline in net income. Apple maintained revenue growth and positive operating cash flow growth, although net income and profit margin declined slightly.

These indicators provide a structured view of revenue growth, profitability, balance-sheet changes, and operating cash generation that can be used as inputs for further AI-driven financial analysis.
